Each mass point was generated with 50,000 events. We calculate the efficiencies for each mass point

1) After snapshotting 

2) After selection

```python
points = {
    "1300":[200],
    "1500":[400],
    "2000":[400,800],
    "3000":[800]    
}
```

```python
grid = [(x, str(y)) for x, ys in points.items() for y in ys]
```

```python
grid
```

## 1) after snapshotting

```python
t = ROOT.TFile.Open('../trijet_nano_files/snapshots/HWWsnapshot_MX_3000_MY_800_18_9of10.root')
```

```python
evts = t.Get('Events')
```

```python
type(evts.GetEntriesFast())
```

```python
snaps = glob.glob('../trijet_nano_files/snapshots/HWWsnapshot_MX_*.root')
```

```python
for pt in grid:
    print('mX = {}, mY = {}'.format(pt[0],pt[1]))
    nEvents = 0
    for fName in snaps:
        if (pt[0] in fName) and (pt[1] in fName):
            tF = ROOT.TFile.Open(fName)
            nE = tF.Get("Events").GetEntriesFast()
            nEvents += nE
            tF.Close()
        else:
            continue
    print('    Original: 50000')
    print('    Snapshot: {}'.format(nEvents))
    print('    Fraction: {}'.format(nEvents/50000.))
    print('    Percent:  {}%'.format((nEvents/50000.)*100))
```

## 2) After selection

```python
ls rootfiles/ | grep MX
```

```python
sels = glob.glob('rootfiles/XHYbbWWselection_MX*.root')
```

```python
blah = ROOT.TFile.Open('rootfiles/XHYbbWWselection_MX_3000_MY_800_18.root')
```

```python
blah.ls()
```

```python
hh = blah.Get('MXvMY_particleNet_SR_loose__nominal')
```

```python
hh.Integral()
```

```python
for pt in grid:
    print('mX = {}, mY = {}'.format(pt[0],pt[1]))
    nEvents = 0
    for fName in sels:
        if (pt[0] in fName) and (pt[1] in fName) and ('up' not in fName) and ('down' not in fName):
            tF = ROOT.TFile.Open(fName)
            hist = tF.Get("MXvMY_particleNet_SR_pass__nominal")
            nE = round(hist.Integral())
            nEvents += nE
            tF.Close()
        else:
            continue
            
    print('    Original:  50000')
    print('    Selection: {}'.format(nEvents))
    print('    Fraction:  {}'.format(nEvents/50000.))
    print('    Percent:   {}%'.format((nEvents/50000.)*100))
```

^ This is for the SR pass
